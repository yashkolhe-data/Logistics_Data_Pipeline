```python
gold_dataframes = {}
for folder in os.listdir(silver_path):
    folder_path = os.path.join(silver_path, folder)
    if os.path.isdir(folder_path):
        file = os.listdir(folder_path)[0]
        file_path = os.path.join(folder_path, file)
        df = pd.read_csv(file_path)
        gold_dataframes[folder] = df
print(gold_dataframes.keys())
```

```python
driver_performance = pd.merge(
    drivers,
    driver_metrics,
    on="driver_id",
    how="left"
)
driver_performance.head()
```

```python
route_summary = loads.merge(routes, on="route_id", how="left")

route_summary = route_summary.groupby("route_id").agg({
    "load_id": "count",
    "revenue": "sum",
    "typical_distance_miles": "mean"
}).reset_index()

route_summary.rename(columns={
    "load_id": "total_loads",
    "revenue": "total_revenue",
    "typical_distance_miles": "average_distance"
}, inplace=True)

route_summary.head()
```

```python
import pandas as pd

customer_summary = pd.read_csv(
    "../Gold/customer_revenue_summary/customer_revenue_summary.csv"
)

driver_summary = pd.read_csv(
    "../Gold/driver_performance_summary/driver_performance_summary.csv"
)

route_summary = pd.read_csv(
    "../Gold/route_summary/route_summary.csv"
)

truck_summary = pd.read_csv(
    "../Gold/truck_summary/truck_summary.csv"
)

print("Customer Summary:", customer_summary.shape)
print("Driver Summary:", driver_summary.shape)
print("Route Summary:", route_summary.shape)
print("Truck Summary:", truck_summary.shape)
```

```python
# Gold Layer Data Quality Validation

print("=== Gold Layer Data Quality Check ===")

# 1. Duplicate check
print("\nDuplicate Records:")
print("Customer Summary:", customer_summary.duplicated().sum())
print("Driver Summary:", driver_summary.duplicated().sum())
print("Route Summary:", route_summary.duplicated().sum())
print("Truck Summary:", truck_summary.duplicated().sum())

# 2. Null check
print("\nNull Values:")
print("Customer Summary:")
print(customer_summary.isnull().sum())

print("\nDriver Summary:")
print(driver_summary.isnull().sum())

print("\nRoute Summary:")
print(route_summary.isnull().sum())

print("\nTruck Summary:")
print(truck_summary.isnull().sum())
```

```python
# Replace missing driver performance metrics with 0

driver_summary["average_mpg"] = driver_summary["average_mpg"].fillna(0)

driver_summary["on_time_delivery_rate"] = driver_summary["on_time_delivery_rate"].fillna(0)

driver_summary["average_idle_hours"] = driver_summary["average_idle_hours"].fillna(0)

print("Missing values replaced with 0 successfully!")
```

```python
print(driver_summary[[
    "average_mpg",
    "on_time_delivery_rate",
    "average_idle_hours"
]].isnull().sum())
```

```python
# Gold Layer Business Validation Checks

print("=== Gold Layer Business Validation ===")

# Customer Revenue Validation
print("\nCustomer Revenue:")
print("Minimum Revenue:", customer_summary["annual_revenue"].min())
print("Maximum Revenue:", customer_summary["annual_revenue"].max())

# Driver Performance Validation
print("\nDriver Performance:")
print("Minimum Trips:", driver_summary["trips_completed"].min())
print("Maximum Trips:", driver_summary["trips_completed"].max())

print("Minimum Total Miles:", driver_summary["total_miles"].min())
print("Maximum Total Miles:", driver_summary["total_miles"].max())

# Route Validation
print("\nRoute Performance:")
print("Minimum Loads:", route_summary["total_loads"].min())
print("Maximum Loads:", route_summary["total_loads"].max())

print("Minimum Revenue:", route_summary["total_revenue"].min())
print("Maximum Revenue:", route_summary["total_revenue"].max())

# Truck Validation
print("\nTruck Performance:")
print("Minimum Trips:", truck_summary["trips_completed"].min())
print("Maximum Trips:", truck_summary["trips_completed"].max())

print("Minimum Revenue:", truck_summary["total_revenue"].min())
print("Maximum Revenue:", truck_summary["total_revenue"].max())
```

```python
# Save updated Driver Summary after null handling

driver_summary.to_csv(
    "../Gold/driver_performance_summary/driver_performance_summary.csv",
    index=False
)

print("✅ Updated Driver Performance Summary saved successfully!")
```

```python
# Final Gold Layer Data Quality Check

print("=== FINAL GOLD LAYER DATA QUALITY CHECK ===")

gold_tables = {
    "Customer Summary": customer_summary,
    "Driver Summary": driver_summary,
    "Route Summary": route_summary,
    "Truck Summary": truck_summary
}

for name, df in gold_tables.items():

    print(f"\n{name}")
    print("-" * 40)

    print("Rows:", len(df))
    print("Columns:", len(df))
    print("Duplicate Rows:", df.duplicated().sum())
    print("Total Null Values:", df.isnull().sum().sum())

print("\n✅ Final Gold Layer Data Quality Check Completed!")
```

```python
import pandas as pd

# Load Gold Layer tables

customer_summary = pd.read_csv(
    "../Gold/customer_revenue_summary/customer_revenue_summary.csv"
)

driver_summary = pd.read_csv(
    "../Gold/driver_performance_summary/driver_performance_summary.csv"
)

route_summary = pd.read_csv(
    "../Gold/route_summary/route_summary.csv"
)

truck_summary = pd.read_csv(
    "../Gold/truck_summary/truck_summary.csv"
)

print("✅ Gold Layer tables loaded successfully!")

print("\nCustomer Summary:", customer_summary.shape)
print("Driver Summary:", driver_summary.shape)
print("Route Summary:", route_summary.shape)
print("Truck Summary:", truck_summary.shape)
```

```python
# Top 10 Trucks by Revenue

top_trucks = truck_summary.sort_values(
    by="total_revenue",
    ascending=False
).head(10)

print("\nTop 10 Trucks by Revenue:")
display(top_trucks)
```

```python
# Top 10 Trucks by Utilization Rate

top_utilized_trucks = truck_summary.sort_values(
    by="utilization_rate",
    ascending=False
).head(10)

print("\nTop 10 Trucks by Utilization Rate:")
display(top_utilized_trucks)
```

```python
# Top 10 Trucks by Maintenance Cost

top_maintenance_trucks = truck_summary.sort_values(
    by="maintenance_cost",
    ascending=False
).head(10)

print("\nTop 10 Trucks by Maintenance Cost:")
display(top_maintenance_trucks)
```

```python
# Overall Business KPIs

print("=== OVERALL BUSINESS KPIs ===")

print("\nCustomer Metrics")
print("Total Customers:", len(customer_summary))
print("Total Customer Revenue:",
      customer_summary["annual_revenue"].sum())

print("\nDriver Metrics")
print("Total Drivers:", len(driver_summary))
print("Total Trips Completed:",
      driver_summary["trips_completed"].sum())
print("Total Miles:",
      driver_summary["total_miles"].sum())

print("\nRoute Metrics")
print("Total Routes:", len(route_summary))
print("Total Loads:",
      route_summary["total_loads"].sum())
print("Total Route Revenue:",
      route_summary["total_revenue"].sum())

print("\nTruck Metrics")
print("Total Trucks:", len(truck_summary))
print("Total Truck Trips:",
      truck_summary["trips_completed"].sum())
print("Total Truck Revenue:",
      truck_summary["total_revenue"].sum())
```

```python
#Business Insight Summary
# Top Customer
top_customer = customer_summary.loc[
    customer_summary["annual_revenue"].idxmax()
]

print("\n🏆 Top Customer by Revenue:")
print("Customer ID:", top_customer["customer_id"])
print("Revenue:", top_customer["annual_revenue"])

# Top Driver
top_driver = driver_summary.loc[
    driver_summary["total_revenue"].idxmax()
]

print("\n🏆 Top Driver by Revenue:")
print("Driver ID:", top_driver["driver_id"])
print("Revenue:", top_driver["total_revenue"])

# Top Route
top_route = route_summary.loc[
    route_summary["total_revenue"].idxmax()
]

print("\n🏆 Top Route by Revenue:")
print("Route ID:", top_route["route_id"])
print("Revenue:", top_route["total_revenue"])

# Top Truck
top_truck = truck_summary.loc[
    truck_summary["total_revenue"].idxmax()
]

print("\n🏆 Top Truck by Revenue:")
print("Truck ID:", top_truck["truck_id"])
print("Revenue:", top_truck["total_revenue"])

# Most Utilized Truck
top_utilized_truck = truck_summary.loc[
    truck_summary["utilization_rate"].idxmax()
]

print("\n🚛 Most Utilized Truck:")
print("Truck ID:", top_utilized_truck["truck_id"])
print("Utilization Rate:", top_utilized_truck["utilization_rate"])
```

```python
#Business Insights Table

business_insights = pd.DataFrame({
    "metric": [
        "Top Customer by Revenue",
        "Top Driver by Revenue",
        "Top Route by Revenue",
        "Top Truck by Revenue",
        "Most Utilized Truck"
    ],
    "entity_id": [
        top_customer["customer_id"],
        top_driver["driver_id"],
        top_route["route_id"],
        top_truck["truck_id"],
        top_utilized_truck["truck_id"]
    ],
    "value": [
        top_customer["annual_revenue"],
        top_driver["total_revenue"],
        top_route["total_revenue"],
        top_truck["total_revenue"],
        top_utilized_truck["utilization_rate"]
    ]
})

display(business_insights)
```
